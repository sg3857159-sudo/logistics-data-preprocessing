## Step 1: Import required libraries

```python
import pandas as pd
import numpy as np

print("Python is working successfully!")
print("Pandas version:", pd.__version__)
```

## Step 2: Create our logistics dataset
Dataset intentionally includes missing values (`NaN`) and outliers so the cleaning steps below have something real to fix.

```python
data = {
    "Shipment_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Distance_km": [120, 250, 180, 500, np.nan, 150, 420, 280, 5000, 350],
    "Delivery_Time_hours": [5, 9, 7, 18, 12, np.nan, 15, 10, 22, 13],
    "Shipping_Cost": [500, 900, 700, 1800, 1200, 600, 1600, np.nan, 2200, 1300]
}

df = pd.DataFrame(data)

print("Logistics dataset created successfully!")
df
```

## Step 3: Handle missing values

### Check missing values before cleaning

```python
print("Missing values before cleaning:")
print(df.isnull().sum())
```

### Fill missing numerical values with the median

```python
numeric_columns = df.select_dtypes(include=np.number).columns

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())
```

### Check missing values after cleaning

```python
print("Missing values after cleaning:")
print(df.isnull().sum())
print("Missing values have been handled successfully!")
```

## Step 4: Check the dataset

```python
print("Dataset information:")
print(df.info())
```

```python
print("Missing values:")
print(df.isnull().sum())
```

```python
print("Duplicate rows:")
print(df.duplicated().sum())
```

```python
print("Basic statistics:")
df.describe()
```

## Step 5: Detect outliers using the IQR method

```python
numeric_columns = ["Distance_km", "Delivery_Time_hours", "Shipping_Cost"]

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[(df[column] < lower_limit) | (df[column] > upper_limit)]

    print(f"\n{column}")
    print("Number of outliers:", len(outliers))
    print("Lower limit:", lower_limit)
    print("Upper limit:", upper_limit)
```

### Treat outliers by capping (winsorizing)
Instead of deleting outlier rows and losing data, extreme values are capped at the lower/upper IQR limits.

```python
for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    df[column] = df[column].clip(lower=lower_limit, upper=upper_limit)

print("Outliers capped successfully!")
df
```

## Step 6: Normalize numerical data

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df_normalized = df.copy()

df_normalized[["Distance_km", "Delivery_Time_hours", "Shipping_Cost"]] = scaler.fit_transform(
    df[["Distance_km", "Delivery_Time_hours", "Shipping_Cost"]]
)

print("Normalized data:")
df_normalized.head()
```
